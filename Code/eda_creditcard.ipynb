# EDA, Dataset 1: ULB Credit Card Fraud (creditcard.csv)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 120
plt.rcParams['figure.figsize'] = (10, 6)

DATA_PATH = '../Dataset/creditcardfraud/creditcard.csv'
OUT_DIR = 'outputs/creditcard'
```

```python
df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
print("\nColumns:\n", df.columns.tolist())
```

```python
df.info()
```

```python
print("--- describe ---")
df.describe()
```

```python
print("Missing values per column:")
print(df.isna().sum().sum(), "total missing cells")
```

```python
print("Exact duplicate rows:", df.duplicated().sum())
```

## Class balance

```python
# class balance, this is the headline stat for the whole dataset
fraud_counts = df['Class'].value_counts()
fraud_pct = df['Class'].value_counts(normalize=True) * 100
print(fraud_counts)
print(fraud_pct.round(4))
```

```python
fig, ax = plt.subplots()
sns.countplot(x='Class', data=df, ax=ax)
ax.set_title('Class balance (0 = genuine, 1 = fraud)')
ax.set_yscale('log')
fig.savefig(f'{OUT_DIR}/class_balance.png', bbox_inches='tight')
```

## Amount distribution

```python
# Amount is heavily right-skewed, log scale needed to actually see the distribution
print("Amount skew:", df['Amount'].skew())
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df['Amount'], bins=50, log_scale=(True, False), ax=axes[0])
axes[0].set_title('Amount distribution (log x-axis)')
sns.boxplot(x='Class', y='Amount', data=df, ax=axes[1])
axes[1].set_yscale('log')
axes[1].set_title('Amount by class')
fig.savefig(f'{OUT_DIR}/amount_distribution.png', bbox_inches='tight')
```

## Time distribution

```python
# Time is seconds elapsed since first transaction, covers 2 days
fig, ax = plt.subplots()
df[df['Class'] == 0]['Time'].hist(bins=50, alpha=0.5, label='genuine', density=True, ax=ax)
df[df['Class'] == 1]['Time'].hist(bins=50, alpha=0.5, label='fraud', density=True, ax=ax)
ax.set_title('Transaction time by class (density)')
ax.set_xlabel('Seconds since first transaction')
ax.legend()
fig.savefig(f'{OUT_DIR}/time_distribution.png', bbox_inches='tight')
```

## Correlations

```python
# correlation heatmap, V1-V28 are PCA components so should be close to orthogonal
corr = df.corr()
fig, ax = plt.subplots(figsize=(14, 12))
sns.heatmap(corr, cmap='coolwarm', center=0, ax=ax)
ax.set_title('Correlation matrix')
fig.savefig(f'{OUT_DIR}/correlation_heatmap.png', bbox_inches='tight')
```

```python
# which V features actually move with Class
class_corr = corr['Class'].drop('Class').sort_values(key=abs, ascending=False)
print("Features most correlated with Class:")
class_corr.head(10)
```

```python
top_features = class_corr.head(6).index.tolist()
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for feat, ax in zip(top_features, axes.flat):
    sns.boxplot(x='Class', y=feat, data=df, ax=ax)
    ax.set_title(feat)
fig.tight_layout()
fig.savefig(f'{OUT_DIR}/top_features_by_class.png', bbox_inches='tight')
```
